```python
lijst = [1,2,3,4,5,6]

def gemiddelde(lijst):
    return sum(lijst)/ len(lijst)

def max_min(lijst):
    return max(lijst), min(lijst)

print(gemiddelde(lijst))
print(max_min(lijst))
```

```python
studenten = {}


start = input("Druk 1 om te starten, 2 om te stoppen")

while start != "1" and start != "2":
     print ("ongeldige invoer")
     start = input("Druk 1 om te starten, 2 om te stoppen")
    


if start == "1":
    while True:
        naam = input("Voeg een naam toe")
        while any(letter.isdigit() for letter in naam):
            print("Een naam mag geen cijfers bevatten")
            naam = input("Voeg een naam toe") 
        naam = naam.capitalize()

        leeftijd = input("wat is de leeftijd")
        while not leeftijd.isdigit():
             print("Voer de leeftijd als cijfer in")
             leeftijd = input("wat is de leeftijd")
             
        leeftijd = int(leeftijd)
        
        studenten[naam] = leeftijd

        
        
        start = input("druk 1 om verder te gaan, 2 om te stoppen")
        while start !="1" and start !="2":
            print ("ongeldige invoer")
            start = input("Druk 1 om te starten, 2 om te stoppen")

        if start == "2":
             break
        

with open("student.txt", "w") as Namenbestand:
        for naam1, leeftijd in studenten.items():
            Namenbestand.write(f"{naam1}: {leeftijd}\n")

with open("student.txt", "r") as Namenbestand:
    inhoud = Namenbestand.readlines()
for regel in inhoud:
    print(regel.strip())

else: print("Gestopt")
    
```
